```python
print("--==Stack dengan Iist==--\n")

stack = []

# Fungsi append() untuk memasukkan (push) elemen ke dalam stack
stack.append('a')
stack.append('i')
stack.append('u')
stack.append('e')
stack.append('o')

print('Initial stack')
print(stack)

# Fungsi pop() untuk mengambil (pop) elemen dari stack dalam urutan LIFO
print('\nElements popped from stack:')
print(stack.pop())
print(stack.pop())
print(stack.pop())

print('\nStack after elements are popped:')
print(stack)
```

```python
print("--==Stack dengan Collections.deque==--\n")

from collections import deque

stack = deque()

stack.append('a')
stack.append('i')
stack.append('u')
stack.append('e')
stack.append('o')

print("initial stack:\n")
print(stack)

print("\nelements popped from stack")
print(stack.pop())
print(stack.pop())
print(stack.pop())

print("\nstack after element are popped:")
print(stack)
```

```python
print("--==Stack dengan queue.LifoQueue==--\n")

from queue import LifoQueue

stack = LifoQueue(maxsize=5)

# qsize() menunjukkan jumlah elemen dalam stack
print(stack.qsize())

stack.put('a')
stack.put('i')
stack.put('u')
stack.put('e')
stack.put('o')

print("Full: ", stack.full())
print("Size: ", stack.qsize())

# fungsi get() untuk mengambil (pop) elemen dari stack dalam urutan LIFO
print('\nElements popped from the stack')
print(stack.get())
print(stack.get())
print(stack.get())

print("\nEmpty: ", stack.empty())
```

```python
print("--==Stack dengan single Linked List==--\n")

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class Stack:
    # Menginisialisasi stack, Gunakan dummy node untuk mempermudah penanganan kasus tepi.
    def __init__(self):
        self.head = Node("head")
        self.size = 0

    # Representasi string dari stack
    def __str__(self):
        cur = self.head.next
        out = ""
        while cur:
            out += str(cur.value) + "->"
            cur = cur.next
        return out[:-2]

    # Mendapatkan ukuran stack saat ini
    def getSize(self):
        return self.size

    # Memeriksa apakah stack kosong
    def isEmpty(self):
        return self.size == 0

    # Mendapatkan item teratas dari stack
    def peek(self):
        # Pemeriksaan keamanan jika kita mengintip stack yang kosong.
        if self.isEmpty():
            raise Exception("Peeking from an empty stack")
        return self.head.next.value

    # Memasukkan (push) nilai ke dalam stack.
    def push(self, value):
        node = Node(value)
        node.next = self.head.next
        self.head.next = node
        self.size += 1

    # Menghapus nilai dari stack dan mengembalikannya.
    def pop(self):
        if self.isEmpty():
            raise Exception("Popping from an empty stack")
        remove = self.head.next
        self.head.next = self.head.next.next
        self.size -= 1
        return remove.value

# Driver Code
if __name__ == "__main__":
    stack = Stack()
    for i in range(1, 11):
        stack.push(i)
    print(f"Stack: {stack}")

    for _ in range(1, 6):
        remove = stack.pop()
        print(f"Pop: {remove}")
    print(f"Stack: {stack}")
```
