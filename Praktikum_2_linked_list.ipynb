```python
print("Penghapusan node diakhir")

class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Fungsi baru untuk menelusuri dan mencetak list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # FUNGSI UTAMA: Menghapus node terakhir
    def RemoveEnd(self):
        last = self.headval
        while (last is not None):
            # Cek apakah ini gerbong terakhir
            if last.nextval == None:
                break
            # Simpan gerbong saat ini sebagai 'sebelumnya' sebelum maju
            prev = last
            last = last.nextval

        if (last == None):
            return

        # Putuskan sambungan gerbong sebelum terakhir
        prev.nextval = last.nextval
        last = None

# Membuat data
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

# Membuat Linked list
Li = LinkedList()

# 3. Menghubungkan node
Li.headval = n1          # head
Li.headval.nextval = n2  # n1 nyambung ke n2
n2.nextval = n3          # n2 nyambung ke n3

print("Sebelum dilakukan remove:")
Li.listprint()
print("-" * 30)

# Menghapus gerbong terakhir (Maret)
Li.RemoveEnd()

print("Setelah dilakukan remove node terakhir:")
Li.listprint()
```

```python
print("---Menghapus node dengan kata kunci\n")
print("==========================================")

class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Fungsi baru untuk menelusuri dan mencetak list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

# Fungsi Penghapusan berdasarkan Kunci (Data)
    def RemoveNode(self, Removekey):
        HeadVal = self.headval

        # Cek jika head adalah yang ingin dihapus
        if (HeadVal is not None):
            if (HeadVal.dataval == Removekey):
                # Catatan: Di gambar tertulis HeadVal.next,
                # namun secara konsisten harusnya HeadVal.nextval
                self.headval = HeadVal.nextval
                HeadVal = None
                return

        # Mencari node yang akan dihapus
        while (HeadVal is not None):
            if HeadVal.dataval == Removekey:
                break
            prev = HeadVal
            HeadVal = HeadVal.nextval

        # Jika kunci tidak ditemukan
        if (HeadVal == None):
            return

        # Menghubungkan node sebelum dan sesudah node yang dihapus
        prev.nextval = HeadVal.nextval
        HeadVal = None

# Membuat data
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

# Membuat Linked list
Li = LinkedList()

# 3. Menghubungkan node
Li.headval = n1          # head
Li.headval.nextval = n2  # n1 nyambung ke n2
n2.nextval = n3          # n2 nyambung ke n3

# Menghapus "Maret"
Li.RemoveNode("Februari")

# 4. Mencetak seluruh isi list dengan satu perintah
Li.listprint()
```

```python
print("---Penyisipan diakhir\n")
print('================================')

class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Fungsi baru untuk menelusuri dan mencetak list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # Fungsi untuk menambah node di akhir
    def AddInEnd(self, newdata):
        NewNode = Node(newdata)
        if self.headval is None:
            self.headval = NewNode
            return
        last = self.headval
        while(last.nextval):
            last = last.nextval
        last.nextval = NewNode

# Membuat data
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

# Membuat Linked list
Li = LinkedList()

# 3. Menghubungkan node
Li.headval = n1          # head
Li.headval.nextval = n2  # n1 nyambung ke n2
n2.nextval = n3          # n2 nyambung ke n3

# --- Menjalankan Fungsi ---
# Menambah "The Last" di urutan paling akhir
Li.AddInEnd("The Last")

# 4. Mencetak seluruh isi list dengan satu perintah
Li.listprint()
```

```python
print("---Penysipan ditengah\n")
print("==============================")

class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class LinkedList:
    def __init__(self):
        self.headval = None

    # Fungsi baru untuk menelusuri dan mencetak list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # Fungsi penyisipan di tengah
    def AddInbetween(self, mid_node, newdata):
        if mid_node is None:
            print("The mentioned node is absent")
            return

        NewNode = Node(newdata)
        # Gerbong baru menunjuk ke node setelah mid_node
        NewNode.nextval = mid_node.nextval
        # mid_node menunjuk ke gerbong baru
        mid_node.nextval = NewNode

# Membuat data
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

# Membuat Linked list
Li = LinkedList()

# 3. Menghubungkan node
Li.headval = n1          # head
Li.headval.nextval = n2  # n1 nyambung ke n2
n2.nextval = n3          # n2 nyambung ke n3

# --- Melakukan Penyisipan ---
# Menyisipkan "Middle" setelah n1 (Januari)
Li.AddInbetween(n1, "Middle")

# 4. Mencetak seluruh isi list dengan satu perintah
Li.listprint()
```

```python
print("--==LATIHAN==--")
print("1. Buatlah double linked list dengan 3 node didalamnya, kemudian lakukan penyisipan tengah dan hapus dengan kata kunci !\n")

class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None
        self.prevval = None

class DoubleLinkedList:
    def __init__(self):
        self.headval = None


    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # Tambah data di tengah
    def AddInbetween(self, mid_node, newdata):
        if mid_node is None:
            return

        NewNode = Node(newdata)
        NewNode.nextval = mid_node.nextval
        NewNode.prevval = mid_node


        if mid_node.nextval is not None:
            mid_node.nextval.prevval = NewNode

        mid_node.nextval = NewNode

    # Hapus pakai kata kunci
    def RemoveNode(self, Removekey):
        HeadVal = self.headval

        if (HeadVal is not None):
            if (HeadVal.dataval == Removekey):
                self.headval = HeadVal.nextval
                if self.headval is not None:
                    self.headval.prevval = None
                HeadVal = None
                return


        while (HeadVal is not None):
            if HeadVal.dataval == Removekey:
                break
            HeadVal = HeadVal.nextval

        if (HeadVal == None):
            return

        if HeadVal.nextval is not None:
            HeadVal.nextval.prevval = HeadVal.prevval

        if HeadVal.prevval is not None:
            HeadVal.prevval.nextval = HeadVal.nextval

        HeadVal = None

dll = DoubleLinkedList()
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

dll.headval = n1
n1.nextval = n2
n2.prevval = n1
n2.nextval = n3
n3.prevval = n2

print("Awal:")
dll.listprint()

# Tugas: Sisip tengah
dll.AddInbetween(n2, "SISIPAN")
print("\nSetelah Sisip:")
dll.listprint()

# Tugas: Hapus pakai kunci
dll.RemoveNode("Februari")
print("\nSetelah Hapus:")
dll.listprint()
```

```python
print("--==LATIHAN==--")
print("2. Buatlah circular linked list dengan 3 node didalamnya, kemudian lakukan penyisipan akhir dan hapus akhir !\n ")



class Node:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None

class CircularLinkedList:
    def __init__(self):
        self.headval = None

    def listprint(self):
        printval = self.headval
        if self.headval is not None:
            while True:
                print(printval.dataval)
                printval = printval.nextval
                if (printval == self.headval):
                    break

    # Tugas: Penyisipan Akhir
    def AddInEnd(self, newdata):
        NewNode = Node(newdata)
        if self.headval is None:
            self.headval = NewNode
            NewNode.nextval = self.headval
            return

        last = self.headval
        while(last.nextval != self.headval):
            last = last.nextval

        last.nextval = NewNode
        NewNode.nextval = self.headval

    # Tugas: Hapus Akhir
    def RemoveEnd(self):
        if self.headval is None:
            return

        last = self.headval
        if last.nextval == self.headval:
            self.headval = None
            return

        while (last.nextval != self.headval):
            prev = last
            last = last.nextval

        prev.nextval = self.headval
        last = None

cll = CircularLinkedList()
cll.AddInEnd("Satu")
cll.AddInEnd("Dua")
cll.AddInEnd("Tiga")

print("Circular Awal:")
cll.listprint()

# Sisip Akhir
cll.AddInEnd("Empat")
print("\nSetelah Sisip Akhir:")
cll.listprint()

# Hapus Akhir
cll.RemoveEnd()
print("\nSetelah Hapus Akhir:")
cll.listprint()
```
