```python
print("--==Graph dengan adjacency list==--")

#Membuat graph tak berarah menggunakan adjacency list
graph = {
    'A' : ['B' ,'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E']
}

#Menampilkan graph
def print_graph(graf):
  for node in graf:
    print(f"{node} -> {graf[node]}")

print_graph(graph)
```

```python
print("--==TRAVERSAL GRAPH MENGGUNAKAN BFS==--")

from collections import deque

graph = {
    'A' : ['B' ,'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E']
}

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex]if neighbor not in visited])

#contoh pemanggilan
bfs(graph, 'B')
```

```python
print("--==Traversal graph menggunakan==--")

graph = {
    'A' : ['B' ,'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E']
}

def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

#pemanggilan
dfs(graph, 'F')
```

```python
print("--==menghitung panjang daftar neighbor==--")

graph = {
    'A' : ['B' ,'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E']
}

def count_neighbors(graph):
  result = {}
  for node in graph:
    result[node] = len(graph[node])

  return result

hasil = count_neighbors(graph)
print(hasil)
```

```python
print("--===T U G A S===--\n")

from collections import deque

# SOAL 1: Graf baru (5 simpul, 7 sisi) & fungsi print_graph

graph = {
    '1' : ['2', '4'],
    '2' : ['1', '5',],
    '3' : ['4', '5'],
    '4' : ['2', '3', '5'],
    '5' : ['3', '4']
}



def tampilkan_graph(graf):
  for i in graf:
    print(f"{i} => {graf[i]}")

print("--> Soal 1")
tampilkan_graph(graph)
print("-"*30, '\n')

# SOAL 2 : Implementasi BFS dan DFS Standar

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex]if neighbor not in visited])
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()
  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

print("--> Soal 2")
print("Traversal BFS (mulai dari 1):", end=" ")
bfs(graph, '1')
print("\nTraversal DFS (mulai dari 1):", end=" ",)
dfs(graph, '1')
print("\n")
print("-"*30, '\n')

#Soal 3 : Modifikasi bfs agar mengambil list urutan kunjungan
def bfs_modif(graph, start):
  visited = set()
  queue = deque([start])
  urutan = []

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      urutan.append(vertex)
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

  return urutan

print("--> Soal 3")
hasil = bfs_modif(graph, '1')
print("Fungsi BFS berhasil mengemalikan list : ", hasil)
print("-"*30, '\n')

#Soal 4 : fungsi find_parth dfs
def find_path(graph, start, end, path=None):
  if path is None:
    path = []
  path = path + [start]

  if start == end:
    return path
  if start not in graph:
    return None

  for neighbor in graph[start]:
    if neighbor not in path:
      path_baru = find_path(graph, neighbor, end, path)
      if path_baru:
        return path_baru
  return None

print("--> Soal 4")
jalur = find_path(graph, '1', '5')
print("Jalur dari 1 menuju 5 : ", jalur)
print("-"*30, '\n')

#Soal 5 : Fungsi Is_connected(graph) *mengecek apakah seluruh vertex terhubung
def is_connected(graph):
  if not graph:
    return True

  mulai = list(graph.keys())[0]
  visited = set()

  def pembantu(node):
    visited.add(node)
    for neighbor in graph[node]:
      if neighbor not in visited:
        pembantu(neighbor)

  pembantu(mulai)
  return len(visited) ==len(graph)

print('--> Soal 5')
print ("apakah graph ing terhubung? : ", is_connected(graph))
print("-"*100)
```
